# Part 1 👉 SCRAPPING DATA FROM WEBSITE

## Import Library

```python
import requests
import pandas as pd
import os

print("✅ Step 1 complete: Libraries loaded successfully.")
```

## Ping the Live API (Connection Test)

```python
# 🌐 The web endpoint hosting the target real estate dataset
api_url = "https://raw.githubusercontent.com/datasets/house-prices-uk/master/data/data.csv"

print("📡 Pinging server... please wait.")
# Send a quick request to check the server status
response = requests.get(api_url, timeout=10)

# Check the HTTP Response Status Code (200 means everything is perfectly OK)
print(f"Server Response Code: {response.status_code}")

if response.status_code == 200:
    print("🎉 Success! The web endpoint is online and accessible.")
else:
    print("🚨 Warning: The server is unreachable or throwing an error.")
```

## Stream and Read Web Data

```python
print("📥 Streaming data from the cloud endpoint...")

try:
    # ⚙️ We switch to engine='python' and skip structural mismatches
    df_market = pd.read_csv(
        api_url, 
        engine='python', 
        on_bad_lines='skip'
    )
    
    print("\n🎉 Web Importation Successful!")
    print(f"Total Rows Captured: {df_market.shape[0]:,}")
    print(f"Total Columns Discovered: {df_market.shape[1]}")
    
    # Let's inspect the headers to see what property features are available
    print("\n📊 Feature Columns:")
    print(df_market.columns.tolist())
    
except Exception as e:
    print(f"🚨 Parsing still failing due to structural layout: {str(e)}")

```
